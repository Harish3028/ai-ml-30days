```python
# Exercise 1: parse_int

def parse_int(text):
    try:
        return int(text)
    except ValueError:
        return None

print(parse_int("abc")) 
print(parse_int("42"))  # Should print: 42 
```

```python
# Exercise 2: files

with open("data.txt", "w") as file:
    file.write("70\n85\n90\n65")

with open("data.txt", "r") as file:
    numbers = [int(line) for line in file]
    
def average(numbers):
    return sum(numbers) / len(numbers)
print(average(numbers))  # Should print the average of the numbers in the file
```

```python
# Exercise 3: BankAccount

class BankAccount:
    def __init__(self, balance=0):
        self.balance = float(balance)

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be more than 0")
        self.balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal must be more than 0")
        if amount > self.balance:
            raise ValueError("Insufficient funds")
        self.balance -= amount

    def __str__(self):
        return f"BankAccount(balance={self.balance})"

my_account = BankAccount(100)
print(my_account)  # Should print: BankAccount(balance=100.0)

my_account.deposit(50)
my_account.withdraw(30)
print(my_account)  # Should print: BankAccount(balance=120.0)

try:
    my_account.withdraw(200)
except ValueError as e:
    print(f"Caught an exception: {e}")
    
print(f"Final balance: {my_account.balance}")  # Should print: Final balance: 120.0
```
