```python
import numpy as np
#Create temps = np.array([28.5, 31.0, 29.5, 33.0, 30.5]). Print the mean, the max, and the index of the max.
#Convert temps to Fahrenheit (F = C * 9 / 5 + 32) in one line, with no loop.
temp = np.array([28.5, 31.0, 29.5, 33.0, 30.5])
print(temp.mean(), temp.max(), temp.argmax())

F = temp * 9 / 5 + 32
print(F)
```

```python
#Create np.arange(1, 11). Print its squares and the sum of the squares.
arr = np.arange(1,11)
squares = arr ** 2
print(squares)
print(squares.sum())
```
