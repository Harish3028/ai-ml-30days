```python
import numpy as np

price = ([10,20,30])
print([p * 2 for p in price])

arr = np.array([10,20,30])
print(arr * 2)

print(price * 2)
```

```python
#creating arrays and seeded method

a = np.array([1,2,3,4])
print(np.zeros(5))
print(np.ones((2,3)))
print(np.arange(0,10,2))
print(np.linspace(0,1,5))

rng = np.random.default_rng(40)
print(rng.integers(1, 100, size=5))
print(rng.random(3))
```

```python
#data shapes and array float

m = np.array([[1,2,3], 
             [4,5,6]])
print(m.dtype)
print(m.ndim)
print(m.shape)
print(m.size)

print(np.array([1,2,3.5]))
print(np.array([1,2,3])/2)
print(np.array([1,2,3], dtype=float))
```

```python
#Vectorized math

x = np.array([1,2,3,4,5])
print(x + 10)
print(x * x)
print(x ** 2)
print(np.sqrt(x))
print(x > 2)

a = np.array([1,2,3])
b = np.array([10,20,30])
print(a * b)
print(a + b)
```

```python
scores = np.array([85,92,78,90])
print(scores.sum(), scores.mean(), scores.max(), scores.min())
print(scores.std())
print(scores.argmax())
```
