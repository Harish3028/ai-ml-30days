```python
#Create a = np.arange(10, 101, 10). 
# Print the first three elements, the last two, every other element, and the array reversed.
import numpy as np 

a = np.arange(10,101,10)
print(a[:3])
print(a[-2:])
print(a[::2])
print(a[::-1])
```

```python
#Create m = np.arange(1, 17).reshape(4, 4). 
# Print the second column, the bottom-right 2×2 block, and the sum of each row.
m = np.arange(1, 17).reshape(4,4)
print(m)
print("\n")
print(m[:, 1])
print(m.sum(axis=1))
print("\n")
print(m[-2:, -2:])
```

```python
#With scores = np.array([85, 42, 91, 67, 78, 55, 99, 60]), 
# print the scores 80 or above, how many failed (below 60), and the percentage that passed (60 or above)
scores = np.array([85,42,91, 67, 78, 55, 99, 60])
print(scores[scores >= 80])
print((scores < 60).sum())
print((scores >= 60).mean()*100)
```

```python
#Using the grades matrix from Part 10, 
# print each student's best exam score, and the index of the exam with the highest class average.
grades = np.array([[85, 90, 78],
                   [70, 65, 80],
                   [92, 88, 95]])
print(grades.max(axis=1))
print(grades.mean(axis=0).argmax())
```
