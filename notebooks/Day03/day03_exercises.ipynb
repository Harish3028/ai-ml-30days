```python
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "name":   ["Aisha", "Bob", "Cy", "Dana", "Eli", "Farah"],
    "course": ["AI", "AI", "Data", "Data", "AI", "Data"],
    "quiz":   [85, 70, 90, 100, 55, 88],
    "final":  [88, 72, 91, 97, 60, 86],
})
```

```python
#Print the shape, the column names, and the mean of final.
print(df.shape)
print(df.columns)
print(df["final"].mean())
```

```python
#Select only the name and final columns.
print(df[["name", "final"]])
```

```python
#Print the names of students in the Data course with final 90 or above.
print(df.loc[(df["course"] == "Data") & (df["final"] >= 90), "name"].tolist())
```

```python
#Add an average column (the quiz and final average), then sort by it from highest to lowest.
df["average"] = (df["quiz"] + df["final"]) / 2
print(df.sort_values("average", ascending=False))
```

```python
#Show the students in the AI course with final 70 or above, and count how many students are in each course.
print(df[(df["course"] == "AI") & (df["final"] >= 70)])
print(df["course"].value_counts())
```
