```python
import pandas as pd
import numpy as np

s = pd.Series([85, 70, 90], index=["Aisha", "Bob", "Cy"])
print(s)
print(s["Bob"])
print(s * 2)
print(s.mean())
```

```python
df = pd.DataFrame({
    "name":   ["Aisha", "Bob", "Cy", "Dana", "Eli", "Farah"],
    "course": ["AI", "AI", "Data", "Data", "AI", "Data"],
    "quiz":   [85, 70, 90, 100, 55, 88],
    "final":  [88, 72, 91, 97, 60, 86],
})
print(df)
print(df.head(3))
print(df.shape)
print(df.dtypes)
df.info()
print(df.describe())
print(df["final"])
print(type(df["final"]))
print(df[["name", "final"]])
```

```python
print(df["final"])
print(type(df["final"]))
print(df[["name", "final"]])
```

```python
mask = df["final"] >= 80
print(mask)
print(df[mask])
print(df[df["final"] >= 80])
```

```python
print(df[(df["course"] == "AI") & (df["final"] >= 70)])
```

```python
print(df[df["course"].isin(["AI"])])
```

```python
print((df["final"] >= 80).sum())
print((df["course"].value_counts()))
```

```python
print(df.loc[df["final"] >= 80, "name"].tolist())
```

```python
#print(df.sort_values("final"))
#print(df.sort_values("final", ascending=False))
print(df.sort_values(["course", "final"], ascending=[True, False]))
```

```python
df["average"] = (df["quiz"] + df["final"]) / 2
df["grade"] = np.where(df["average"] >= 80, "A",
                       np.where(df["average"] >= 60, "B", "F"))
print(df)
```
