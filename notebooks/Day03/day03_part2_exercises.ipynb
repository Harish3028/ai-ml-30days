```python
import pandas as pd
import numpy as np

# Exercises 1, 2, 3
df = pd.DataFrame({
    "name":   ["Aisha", "Bob", "Cy", "Dana", "Eli", "Farah"],
    "course": ["AI", "AI", "Data", "Data", "AI", "Data"],
    "quiz":   [85, 70, 90, 100, 55, 88],
    "final":  [88, 72, 91, 97, 60, 86],
})
```

```python
#GroupBy: the average quiz and average final per course, in one table.
print(df.groupby("course")[["quiz", "final"]].mean())
```

```python
#Named aggregation: per course, the number of students and the best final.
df2 = df.groupby("course").agg(students=("name" ,"count"), best=("final" ,"max"))
print(df2)
```

```python
#Merge: do a left merge of df with courses on course, and print name, course, lecturer. 
#Then run the same merge with how="outer" and say how many rows it has and what's different.
courses = pd.DataFrame({
    "course":   ["AI", "Data", "Cloud"],
    "lecturer": ["Dr. Lim", "Dr. Tan", "Dr. Wong"],
    "credits":  [4, 3, 2],
})

df3 = df.merge(courses, on="course", how="left")
print(df3[["name", "course", "lecturer"]])
```

```python
df3 = df.merge(courses, on="course", how="outer")
print(df3[["name", "course", "lecturer"]])
```

```python
#Missing values: with the m table from Part 3 (re-create it, since you changed it above), 
#print the missing count per column, then fill the missing quiz values with the mean, 
#then drop the rows where final is missing.
m = pd.DataFrame({
    "name":  ["Aisha", "Bob", "Cy", "Dana", "Eli"],
    "quiz":  [85, np.nan, 90, 100, np.nan],
    "final": [88, 72, np.nan, 97, 60],
})
print(m.isna().sum())
m["quiz"] = m["quiz"].fillna(m["quiz"].mean())
print(m["quiz"].tolist())

print(m.dropna(subset=["final"]))
```
