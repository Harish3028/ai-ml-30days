- probability = (outcomes we want) / (all outcomes)
- Not: P(not A) = 1 - P(A)
- And (for independent events): P(A and B) = P(A) * P(B)
- Or: at least one of A or B happens. Easiest way: 1 - P(neither)

```python
import numpy as np 
import pandas as pd

rng = np.random.default_rng(0)

for n in [10, 100, 1000]:
    rolls = rng.integers(1, 7, size=n)
    print(n, round((rolls == 6).mean(), 4))
```

```python
rng = np.random.default_rng(1)
d1 = rng.integers(1, 7, size=100000)
d2 = rng.integers(1, 7, size=100000)
print(round(((d1 == 6) & (d2 == 6)).mean(), 4))
print(round(((d1 == 6) | (d2 == 6)).mean(), 4))
```
