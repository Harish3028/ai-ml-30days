```python
import numpy as np
import pandas as pd

a = np.array([48, 49, 50, 51, 52])
b = np.array([10, 30, 50, 70, 90])

dev_a = a - a.mean()
print(dev_a)
print(dev_a.sum())
print(dev_a ** 2)

val_a = (dev_a ** 2).mean()
print(val_a)
print(np.sqrt(val_a))
```

```python
dev_b = b - b.mean()
print(dev_b)
print(dev_b.sum())
print(dev_b ** 2)

val_b = (dev_b ** 2).mean()
print(val_b)
print(np.sqrt(val_b))
print(pd.Series(b).var())
```

```python
print(np.var(a), np.std(a))
print(pd.Series(a).var(), pd.Series(a).std())
```

```python
a = pd.Series([48, 49, 50, 51, 52])
o = pd.Series([48, 49, 50, 51, 152])

for name, s in [("a", a), ("o", o)]:
    iqr = s.quantile(0.75) - s.quantile(0.25)
    print(name, s.mean(), s.median(), round(s.std(), 2), iqr)
```

```python
def count_outliers(series):
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    low = q1 - 1.5 * iqr
    high = q3 + 1.5 * iqr
    outside = (series < low) | (series > high)
    return int(outside.sum())
```

```python
print(count_outliers(a))
print(count_outliers(o))
```

```python
scores = np.array([35, 42, 48, 50, 55, 55, 58, 60, 62, 65,
                   68, 70, 72, 75, 78, 80, 85, 90, 92, 95])

print(np.percentile(scores, [25,50,75]))
print(scores.mean())
```

```python
below = scores < 78
print(below.sum())
print(below.mean() * 100)
```

```python
def percentile_rank(scores, value):
    below = scores < value
    return below.mean() * 100
```

```python
print(percentile_rank(scores, 78))     # expected 70.0
print(percentile_rank(scores, 35))     # expected 0.0
print(percentile_rank(scores, 100))    # expected 100.0
```

```python
q1 = np.percentile(scores, 25)
q3 = np.percentile(scores, 75)
iqr = q3 - q1
low = q1 - 1.5 * iqr
high = q3 + 1.5 * iqr
print(q1, q3, iqr, low, high)

outliers = scores[(scores < low) | (scores > high)]
print(outliers)
```

```python
x = np.array([1, 2, 3, 4, 5])          # hours studied
y = np.array([50, 55, 65, 70, 85])     # exam score

dev_x = x - x.mean()
dev_y = y - y.mean()
print(dev_x)
print(dev_y)
print(dev_x * dev_y)
cov_xy = (dev_x * dev_y).mean()
print(cov_xy)
```

```python
print(np.cov(x, y))
print(np.cov(x, y)[0, 1])
print(np.corrcoef(x, y)[0, 1])
```

```python
corr_xy = cov_xy / (x.std() * y.std())
print(corr_xy)
```
