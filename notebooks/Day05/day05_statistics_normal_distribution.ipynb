```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rng = np.random.default_rng(42)
heights = rng.normal(loc=170, scale=10, size=10000)

print(round(heights.mean(), 2), round(heights.std(), 2))
print(round(heights.min(), 1), round(heights.max(), 1))

plt.hist(heights, bins=50)
plt.title("Simulated heights (cm)")
plt.xlabel("Height (cm)")
plt.ylabel("Number of people")
plt.show()
```

```python
mean = heights.mean()
std = heights.std()

for k in [1,2,3]:
    inside = (np.abs(heights - mean) < k * std).mean() * 100
    print(k, round(inside, 1))
```

```python
z = (heights - mean) / std
print((np.abs(z) > 3).sum())
```
