```python
import numpy as np

# 1. shapes
v = np.array([2, 3, 4])
M = np.array([[1, 2, 3],
              [4, 5, 6]])
print(v.shape, M.shape, M.T.shape)          # M.T = transpose (swap rows and columns)

# 2. dot product, three ways
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.sum(a * b), np.dot(a, b), a @ b)

# 3. one prediction = x @ w + b
x = np.array([100, 3, 10])     # area, rooms, age
w = np.array([3, 5, -2])       # weights
bias = 20
print(x @ w + bias)

# 4. many predictions at once: X has one house per row
X = np.array([[100, 3, 10],
              [80, 2, 5],
              [120, 4, 20]])
print(X @ w + bias)

# 5. matrix @ matrix
A = np.array([[1, 2, 3],
              [4, 5, 6]])
B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
print(A @ B)
#print(A @ A.T)
# A @ A   <- run this alone, read the error
```

```python
# 1. derivative = slope, checked with a tiny nudge h
def f(x):
    return x**2

def numeric_derivative(func, x, h=1e-6):
    return (func(x + h) - func(x)) / h

print(round(numeric_derivative(f, 5), 3))# rule 2x -> 6
print(round(numeric_derivative(f, -2), 3))# rule 2x -> -4
print(round(numeric_derivative(f, 0), 3))# bottom of the bowl -> 0
```

```python
# 2. gradient descent in ONE variable: minimise g(x) = (x - 3)^2
def g_slope(x):
    return 2 * (x - 3) # derivative of (x-3)^2

for lr in (0.1, 1.1):
    x = 0.0
    for step in range(1, 61):
        x = x - lr * g_slope(x)
        if step in (1, 2, 3, 10, 60):
            print(lr, step, round(x, 4))
```

```python
# 3. partial derivatives of the loss for ONE data point
def loss(w, b, x, y):
    return(w * x + b - y) ** 2

w, b, x, y = 0.0, 0.0, 3.0, 6.0
h = 1e-6
num_dw = (loss(w + h, b, x, y) - loss(w, b, x, y)) / h  # nudge only w
num_db = (loss(w, b + h, x, y)  - loss(w, b, x, y)) / h # nudge only b

err = w * x - b - y
dw = 2 * err * x
db = 2 * err
print(loss(w, b, x, y))
print(round(num_dw, 3), round(num_db, 3))
print(dw, db)
```

```python
# 4. one gradient descent step
lr = 0.01
w2 = w - lr * dw
b2 = b - lr * db
print(w2, b2, loss(w2, b2, x, y))
```
