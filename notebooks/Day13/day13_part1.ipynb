```python
import torch, numpy as np
print(torch.__version__, "| GPU available:", torch.cuda.is_available())
```

```python
a = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(a)
print(a.shape, a.dtype, a.device, a.ndim, a.numel())
```

```python
print(torch.zeros(2, 3))
print(torch.arange(0, 10, 2))
print(torch.linspace(0, 1, 5))
torch.manual_seed(42)
print(torch.rand(2, 3))
```

```python
print(a.reshape(3, 2))
print(a.T.shape)
af = a.float()
print(af.dtype, af.mean())
print(a.mean())
```

```python
v = torch.tensor([10, 20, 30])
print(a + v)
print(a * 2)
b = torch.tensor([[1, 0], [0, 1], [1, 1]])
print(a @ b)
print(a.sum(), a.sum(dim=0), a.sum(dim=1))
print(af.mean(dim=0))
```

```python
print(a @ a)
```

```python
print(a[0])
print(a[:, 1])
print(a[1, 2])
print(a[a > 3])
print(a.max(), a.argmax())
```

```python
n = np.array([1.5, 2.5, 3.5])
t = torch.from_numpy(n)
n[0] = 99
print(t)
c = t.clone()
print(c.numpy())
```

```python
x = torch.rand(1000, 1000)
device = "cuda" if torch.cuda.is_available() else "cpu"
x_gpu = x.to(device)
print(x.device, x_gpu.device)
y = x_gpu @ x_gpu
print(y.shape)
```

```python
print(x.to(device) @ x_gpu)
```
