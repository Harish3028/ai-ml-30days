**Part 1: LangChain, the idea**

- What LangChain is

*A library of standard building blocks for LLM apps, with the same interface on every block, so blocks can be swapped and chained. You’ve already built every block by hand, so this will feel familiar. Only the names are new.*

|hat we wrote by hand|LangChain’s name for it
|---|---|
|EvidenceChunk (text plus metadata)|Document
|embed_texts / the embedding model|Embeddings
|EvidenceIndex (FAISS)|Vector store
|EvidenceIndex.search(query, k)|Retriever
|build_user_prompt + judge_v1.txt|Prompt template
|call_llm (Gemini call)|Chat model
|RequirementVerdict.model_validate_json|Output parser / structured output
|judge_with_retrieval (the whole thing)|Chain

question

  * → retriever finds top-k Documents
  * → format them into one context string
  * → prompt template fills {context} and {question}
  * → chat model generates
  * → output parser turns the reply into text or an object
  * → answer

*LangChain connects the blocks with the pipe operator | (called LCEL, the LangChain Expression Language). Every block follows the same Runnable interface (invoke, stream, batch), which is why any two blocks can be joined. The chain is roughly: retriever → format → prompt → model → parser.*

- Why people use it, and when a senior engineer is careful

    * For: swap a model or a vector store by changing one line; many ready-made integrations; tracing tools; and LangGraph, our Day 25 topic, builds on these blocks.

    * Careful: the library changes fast, so tutorials go stale; every layer hides what is really happening; and debugging gets harder if you don’t know what is underneath. That is exactly why we built the pieces by hand first. You can now tell what LangChain does for you and what it hides.
