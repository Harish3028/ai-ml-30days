**LangGraph**

- The problem a chain cannot solve. A chain is a straight line: prompt | model | parser. Real systems need decisions and loops: “if the answer fails validation, try once more; if it still fails, send it to a person.”

- LangGraph is a library for exactly that. You describe the program as a graph:

|Concept|Meaning|Tender example
|---|---|---|
|State|A shared dictionary that every step reads and updates|requirement, evidence, verdict, problems, attempts
|Node|A function that takes the state and returns changes to it|retrieve, judge, validate
|Edge|A fixed arrow: after A, go to B|retrieve → judge → validate
|Conditional edge|A function looks at the state and returns the next node’s name|after validate: retry, stop, or review
|START / END|The entry and exit	
|compile + invoke|Turn the description into a runnable object, then run it with a starting state

**The tender judge as a graph:**

START → retrieve → judge → validate ──ok──────────────→ END

                     ↑          │

                     │      problems and attempts < 2

                     └──────────┘ (retry)

                                │

                          problems and attempts >= 2

                                ↓

                           needs_review → END
