```python
## E1  without using sum() or len(), compute the prices and average of it using a loop
prices = [250000, 310000, 180000, 420000, 295000]
names = ["A", "B", "C", "D", "E"]

total = 0
count = 0
for v in prices:
    total += v
    count += 1
mean = total / count
print(total, mean)
```

```python
## E2  Count how many prices are above 300000 (a loop with an if and a counter).
cnt = 0 
for v in prices:
    if v > 300000:
        cnt += 1
print(cnt)
```

```python
## E3  Build a dictionary {name: price} with zip, loop over .items() to print 
# "A costs 250000", and find the name of the most expensive house WITHOUT using max():
# keep a variable best_name and best_price and update them inside the loop.

build_dict = dict(zip(names, prices))

best_name = None
best_price = -1

for name, price in build_dict.items():
    print(f"name {name} : price {price}")

    if price > best_price:
        best_price = price
        best_name = name

print(f"\nMost expensive house: {best_name} (Costs {best_price})")
```

```python
## E7  Write a function  mean_std(values)  that returns (mean, std) using only loops.
# Call it on prices.  If you are stuck, write STUCK and the line, and send it.

def mean_std(values):
    total = 0
    count = 0
    for v in values:
        total += v
        count += 1
    mean = total / count
    sq = 0
    for v in values:
        sq += (v - mean) ** 2
    std = (sq/count) ** 0.5
    return mean, std
```
