```python
prices = [250000, 310000, 180000, 420000, 295000]
names = ["A", "B", "C", "D", "E"]
```

```python
## E1  Turn F1 into a function. Then call mean(prices)

def mean(values):
    total = 0
    count = 0
    for v in values:
        total += v
        count += 1
    return total/count

mean(prices)
```

```python
## E2  (CORE) A std function with a default argument.  
# ddof means "delta degrees of freedom": the divisor is n minus ddof

def std(values, ddof=0):
    m = mean(values)
    sq = 0
    for v in values:
        sq += (v - m) ** 2
    return (sq / (len(values) - ddof)) ** 0.5
```

```python
print(std(prices)) ## this is the one uses StandarScaler and np.std
print(std(prices, ddof=1)) ## this is the one uses pandas .std()
```

```python
## E4  Comprehensions.  Fill the blanks (the keywords and functions are named):

expensive = [v for v in prices if v> 300000]
price_of = {n: p for n, p in zip(names, prices)}
above = {n: p for n, p in price_of.items() if p > 290000}
count = sum(1 for v in prices if v > 300000)
print(expensive, above, count)
```

```python
## E5  (CORE) FIT AND TRANSFORM.  This is the whole idea of StandardScaler.
def fit(values):
    return mean(values), std(values)

def transform(values, m, s):
    return [(v - m) / s for v in values]
```

```python
train = prices[:4]
test = prices[4:]
m, s = fit(train)
print(m, round(s, 2))
print([round(z, 4) for z in transform(test, m, s)])

## We fit on the train only and reuse the same m and s on test to prevent data leakage
## If the test row helped compute m and s, information from the test data would flow into the model. House E’s z-score was 0.051 using all five prices and became 0.0568 using only the training rows. That is the honest number, because a new house is always scored with numbers learned from the past, never from itself. 
# Leaking would make the test score look better than it truly is. This is also why the scaler goes inside a Pipeline.
```

```python
## E6  From memory, write mean(values), std(values, ddof=0), fit(values) and
# transform(values, m, s), the transform as ONE comprehension line.
# If you are stuck, write STUCK and the line.

def mean(values):
    total = 0
    count = 0
    for v in values:
        total += v
        count += 1
    return total / count

def std(values, ddof=0):
    m = mean(values)
    sq = 0
    for v in values:
        sq += (v - m) ** 2
    return (sq / (len(values) - ddof)) ** 0.5

def fit(values):
    return mean(values), std(values)

def transform(values, m, s):
    return [(v - m) / s for v in values]
```
